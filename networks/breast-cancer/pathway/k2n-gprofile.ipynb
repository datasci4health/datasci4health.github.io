```python
import pandas as pd
from kgml_to_network import KeggKgmlToNetwork
```

```python
# Genes of interest and KEGG pathways
genes_df = pd.read_csv('gprofile_genes.csv')
genes_of_interest = genes_df.iloc[:, 0].tolist()
genes_of_interest = ['hsa:' + str(gene) for gene in genes_of_interest]
pathways_of_interest = ['hsa04151', # PI3K-Akt signaling pathway
                        'hsa04151', # Cytoskeleton in muscle cells
                        'hsa05165'] # Human papillomavirus infection
```

```python
import pandas as pd

# Step 4: Export to CSV
net = KeggKgmlToNetwork()
net.build_from_pathway_and_gene_lists(pathways_of_interest, genes_of_interest)
net.write_cytoscape_csvs("gprofile_cytoscape_nodes.csv", "gprofile_cytoscape_edges.csv")
```
