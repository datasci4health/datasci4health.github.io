```python
from kgml_to_network import KeggKgmlToNetwork
```

```python
conversion_params = {
  "kgml_source": "hsa05224.xml",  # KEGG: Breast cancer pathway
  "csv_full": "hsa05224_gene_fu_pathway.csv",
  "csv_nodes": "hsa05224_cytoscape_nodes.csv",
  "csv_edges": "hsa05224_cytoscape_edges.csv"
}
```

```python
network_builder = KeggKgmlToNetwork(conversion_params["kgml_source"])
```

```python
functional_units = network_builder.extract_functional_units()

print(f"Number of functional units extracted: {len(functional_units)}")
```

## Full Conversion

```python
flat_df = network_builder.write_flat_csv(conversion_params["csv_full"])

flat_df.head()
```

```python
nodes_df, edges_df = network_builder.write_cytoscape_csvs(
    node_path=conversion_params["csv_nodes"],
    edge_path=conversion_params["csv_edges"]
)

nodes_df.head(), edges_df.head()
```
